```python
import os

n = {}
labels = {}

for filename in os.listdir('data/gastemp'):
    if not filename.endswith('.jpg'):
        continue

    ns = filename.replace('.jpg', '').split('_')
    for label in ns:
        if label not in labels:
            labels[label] = 0
        labels[label] += 1
    if len(ns) not in n:
        n[len(ns)] = []
    
    n[len(ns)].append(filename)

for k in n:
    print(f'Files with {k} underscores:', len(n[k]))
    if k == 2:
        for f in n[k]:
            print(f'  {f}')
    print()

# sort labels by alphabetical order
labels = dict(sorted(labels.items()))

print('Labels:')
for label, count in labels.items():
    print(f'  {label}: {count}')
```
